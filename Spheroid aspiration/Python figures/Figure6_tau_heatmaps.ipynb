## Heat Maps of Fitted Retardation Time $\tau$ As a Functionof Cellular Stiffness And Cell-Cell Adhesion

Plots the fitted retardation time $\tau$ across the entire simulated parameter space: $J = -40$, $-45$, $-50$, $-55$, $-60$) and $\lambda_S = 0.8$, $0.9$, $1.0$, $1.1$, $1.2$, for three values of aspiration force.

```python
from pathlib import Path
import numpy as np
import matplotlib.pyplot as plt
from scipy.io import loadmat


def create_heatmap(data, row_labels, col_labels, ax, title, xlabel="", ylabel="", colorbar_label=r"Mean $\tau$"):
    im = ax.imshow(data)
    ax.set_xticks(range(len(col_labels)), col_labels, fontsize=14)
    ax.set_yticks(range(len(row_labels)), row_labels, fontsize=14)
    ax.set_xlabel(xlabel, fontsize=16)
    ax.set_ylabel(ylabel, fontsize=16)
    ax.set_title(title, fontsize=18, pad=10)

    cbar = ax.figure.colorbar(im, ax=ax, pad=0.04, fraction=0.15, shrink=0.72)
    cbar.set_label(colorbar_label, fontsize=14, labelpad=8)
    cbar.ax.tick_params(labelsize=12)

    return im


def get_scores(force):
    base = Path.cwd().parent / "all sims & analysis"
    adhesions = ["60", "55", "50", "45", "40"]
    lambdas = ["0.8", "0.9", "1", "1.1", "1.2"]

    return np.array([
        [
            float(np.squeeze(
                loadmat(
                    base / f"lambda_S_{lam}_J_{adh}" / "sim_analysis_MATLAB" / f"block_{force}_results.mat"
                )[f"mean_tau_{force}"]
            ))
            for lam in lambdas
        ]
        for adh in adhesions
    ])


fig, axes = plt.subplots(1, 3, figsize=(15, 5), constrained_layout=True)

for ax, force, title in zip(
    axes,
    ["06", "08", "1"],
    [r"$\lambda_D = 0.6$", r"$\lambda_D = 0.8$", r"$\lambda_D = 1$"]
):
    create_heatmap(
        get_scores(force),
        ["60", "55", "50", "45", "40"],
        ["0.8", "0.9", "1.0", "1.1", "1.2"],
        ax,
        title,
        xlabel=r"Cellular Stiffness ($\lambda_S$)",
        ylabel=r"Adhesion ($J$)"
    )

for ax in axes[1:]:
    ax.tick_params(labelleft=False)
    ax.set_ylabel("")

plt.show()
```
